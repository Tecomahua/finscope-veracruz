# FinScope Veracruz: infraestructura financiera

Cortes: CNBV 2020–2025; DENUE mayo 2026. Solo municipio 30193. Los universos de ambas fuentes difieren.

```python
from pathlib import Path
import pandas as pd
base = Path('../data/processed')
cnbv = pd.read_csv(base / 'cnbv_acceso.csv', dtype={'municipio_clave':str}, parse_dates=['periodo'])
denue = pd.read_csv(base / 'denue_agregado.csv', dtype={'municipio_clave':str, 'codigo_act':str, 'localidad_clave':str})
assert set(cnbv.municipio_clave)=={'30193'} and set(denue.municipio_clave)=={'30193'}
assert cnbv.periodo.is_unique and (cnbv.adultos>0).all()
cnbv.tail()
```

## Tasas y cambio trimestral
Las tasas utilizan el denominador de adultos incluido por CNBV en el mismo registro. Los cambios de denominador pueden afectar comparaciones.

```python
for col in ['sucursales','corresponsales','cajeros','terminales']:
    cnbv[col + '_por_10k'] = cnbv[col] * 10000 / cnbv.adultos
    cnbv[col + '_cambio'] = cnbv[col].diff()
cnbv.tail(4).round(2)
```

## Directorio DENUE
Conteos agregados por actividad y localidad, sin datos de contacto, nombres o coordenadas. No sumar con CNBV.

```python
denue.groupby(['codigo_act','actividad'], as_index=False).establecimientos.sum().sort_values('establecimientos', ascending=False)
```
